## Functions

```python
# 1. Defining the function
def greet_user(name):          # 'name' is a parameter (input placeholder)
    """This is a docstring.""" # Optional description
    message = f"Hello, {name}!" 
    return message             # Sends the result back to the caller

# 2. Calling the function
result = greet_user("Alice")   # "Alice" is the argument (actual value passed)
print(result)                  # Output: Hello, Alice!
```

```python
def is_even(i):
    """ Optional docstring"""
    x = i%2==0
    return x
is_even(7)
```

### Parameters vs Arguments 

- Default argument
- Positional arguments
- Keyword arguments
- Arbitrary arguments

```python
# 'name' and 'age' are PARAMETERS
def greet(name, age):
    print(f"Hello {name}, you are {age} years old.")

# "Alice" and 25 are ARGUMENTS
greet("Alice", 25)
```

#### Positional Arguments

- Assigned strictly from left to right.
- Order matters entirely.
- Swapping the positions will mismatch data or cause errors.

```python
def describe_pet(animal, name):
    print(f"I have a {animal} named {name}.")

# Correct order
describe_pet("dog", "Buddy")   # Output: I have a dog named Buddy.
# Incorrect order
describe_pet("Buddy", "dog")   # Output: I have a Buddy named dog.
```

### Default Arguments

Default arguments provide a fallback value in the function definition if a parameter is omitted during the call. 
-  Makes passing specific arguments optional.
-  Overridden if a new value is passed.
-  Rule: All default parameters must follow non-default parameters in the definition. 


```python
# 'country' has a default value
def origin(name, country="USA"):
    print(f"{name} is from {country}.")

origin("John")          # Output: John is from USA.
origin("Alex", "India") # Output: Alex is from India.
```

### Keyword Arguments

- You ignore the sequential order entirely.
- Improves code readability.
- Rule: Positional arguments must always come before keyword arguments in a function call.

```python
def describe_pet(animal, name):
    print(f"I have a {animal} named {name}.")

# Order does not matter here
describe_pet(name="Buddy", animal="dog") 
```

### Arbitrary Arguments

1. Non-Keyword Arbitrary Arguments (*args)
   - Captures extra positional arguments into a single tuple.
   - Created using a single asterisk (*) before the variable name

```python
def sum_numbers(*args):
    return sum(args) # 'args' is a tuple: (1, 2, 3, 4)

print(sum_numbers(1, 2, 3, 4, 5)) # Output: 15
```

2. Keyword Arbitrary Arguments (**kwargs)
    - Captures extra named keyword arguments into a single dictionary.
    - Created using double asterisks (**) before the variable name.

```python
def show_profile(**kwargs):
    print(kwargs) # 'kwargs' is a dict: {'user': 'dev12', 'role': 'admin'}

show_profile(user="dev12", role="admin")
```

```python
def flexi(*number):
    product = 1
    print(number)
    print(type(number))
    for i in number:
        product = product * i
    print(product)
```

```python
flexi(1,2,3)
```

```python
flexi(1,2,3,4,5,6,7)
```

### Global Var And Local var

#### Local Variables

```python
# A local variable is created when a function starts and is destroyed as soon as the function finishes executing. 
#If you try to call it outside the function, Python throws a NameError. 


def my_function():
    local_var = "I'm Local variable"
    print(local_var)
my_function()
```

```python
def my_function():
    local_var = "I'm Local variable"
    print(local_var)
my_function()
print(local_var) # This will crash the program.
```

### Global Variables


```python
# A global variable is accessible by any function in your script. 

global_var = "I'm Global Variable"   # Global variable

def my_func():
    print(global_var)   # Functions can naturally read global variables
my_func()
print(global_var)
```

#### Modifying Global Variables (The global Keyword)

```python
score = 10 # global variable

def score_update():
    global score    # Links this function to the global variable
    score = 50     # Modifies the global variable
    print(score)
score_update()
print(score)
```

#### Variable Shadowing

```python
name = "Amit Kaushal"  # Global

def shadow():
    name = "Navneet Kaushal"   # Local shadows the global variable
    print(name)
shadow()
print(name)
```
