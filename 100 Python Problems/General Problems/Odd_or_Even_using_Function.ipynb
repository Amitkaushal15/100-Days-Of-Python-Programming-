```python
def is_even( number ):
    """ Input: i, a positive int
        Return True if i is even, otherwise False
        Created By: Amit Kaushal
        Last edited: 22 July 2026
    """ 
    if number%2==0:
        return"Even"
    else:
        return "Odd"
for i in range(1,11):
    print(is_even(i))
print(is_even.__doc__)
```
